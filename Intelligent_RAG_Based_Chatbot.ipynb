# Split Text into Chunks

This section divides the prepared customer support messages into smaller text chunks for the Retrieval-Augmented Generation (RAG) pipeline. Chunking helps the embedding model represent text in manageable units and allows the retrieval system to find relevant information for a user's question. Since the current documents contain individual tweet messages, we will initially use each message as a document and apply a text splitter to prepare them for embedding and retrieval.

```python
# Split prepared documents into text chunks

from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.create_documents(documents)

print("Documents before chunking:", len(documents))
print("Total chunks created:", len(chunks))

if chunks:
    print("\nFirst chunk:")
    print(chunks[0].page_content)
```

### Observation

The text-splitting process received 9,980 documents and produced 9,980 chunks using a chunk size of 500 characters and a chunk overlap of 50 characters. The first chunk contains the original customer support message requesting the customer to use a private, secure link for further assistance.

### Interpretation

The results indicate that each prepared tweet remained a single chunk because the messages are short relative to the configured chunk size. These chunks can be used in the subsequent embedding and retrieval stages of the RAG pipeline. However, the current results represent only the first 10,000 rows of the dataset, not the complete 5 GB CSV file.

# Create the ChromaDB Vector Store

This section creates a persistent ChromaDB vector store to store the prepared customer support text chunks and their embeddings. The embedding model converts each text chunk into a numerical vector, enabling semantic similarity search when a user asks a question. The vector store will be saved in the project's vector_db directory so it can be reused without recreating the database each time. The initial vector store will contain only the current batch of 9,980 chunks; the full dataset-processing strategy must be completed before treating it as the final knowledge base.

```python
# Initialize the ChromaDB vector store

import chromadb
from pathlib import Path

VECTOR_DB_PATH = Path.cwd() / "vector_db"

VECTOR_DB_PATH.mkdir(parents=True, exist_ok=True)

chroma_client = chromadb.PersistentClient(
    path=str(VECTOR_DB_PATH)
)

collection = chroma_client.get_or_create_collection(
    name="customer_support",
    metadata={"hnsw:space": "cosine"}
)

print("ChromaDB vector store initialized successfully!")
print("Vector database path:", VECTOR_DB_PATH)
print("Collection name:", collection.name)
print("Existing records:", collection.count())
```

### Observation

The ChromaDB vector store was initialized successfully. The persistent database path was set to D:\Intelligent_RAG_Based_Chatbot\vector_db, and the customer_support collection was created or accessed successfully. The collection currently contains 9,980 records.

### Interpretation

The successful initialization confirms that ChromaDB is ready to store customer support text chunks and their embeddings. The zero record count is expected because no documents have been inserted yet. The next step is to generate embeddings for the prepared text chunks and store them in the collection. The current collection represents the initial batch, not the complete dataset.

## Generate Embeddings and Store Documents

This step converts the prepared text chunks into numerical embeddings using the initialized all-MiniLM-L6-v2 model and stores the embeddings, text, and unique identifiers in the existing ChromaDB collection. Batch processing helps manage memory usage during insertion. The current operation processes the 9,980 chunks prepared from the first 10,000 rows of the dataset; it does not yet represent the complete dataset.

```python
# Generate embeddings and store documents in ChromaDB

BATCH_SIZE = 500

for start in range(0, len(chunks), BATCH_SIZE):
    batch_chunks = chunks[start:start + BATCH_SIZE]

    batch_texts = [chunk.page_content for chunk in batch_chunks]
    batch_ids = [
        f"customer_support_{i}"
        for i in range(start, start + len(batch_chunks))
    ]

    batch_embeddings = embedding_model.encode(
        batch_texts,
        show_progress_bar=False
    ).tolist()

    collection.upsert(
        ids=batch_ids,
        documents=batch_texts,
        embeddings=batch_embeddings
    )

print("Documents and embeddings stored successfully!")
print("Total chunks processed:", len(chunks))
print("Records in ChromaDB:", collection.count())
```

### Observation

The embedding generation and document insertion process completed successfully. A total of 9,980 text chunks were processed in batches of 500 and stored in the ChromaDB customer_support collection. The final collection count is 9,980 records.

### Interpretation

The results confirm that the prepared text chunks and their numerical embeddings have been stored in the persistent ChromaDB vector store. The collection is now ready for semantic similarity search. However, these records represent only the first 10,000 rows of the original dataset, so the vector store is not yet the complete knowledge base.

# Test Similarity Search

This section evaluates the retrieval capability of the ChromaDB vector store by searching for customer support messages that are semantically related to a sample query. The query is converted into an embedding using the same model used to embed the stored documents. ChromaDB then retrieves the five most relevant messages based on vector similarity. This step verifies that the stored text and embeddings can support the retrieval component of the RAG pipeline.

```python
# Test semantic similarity search

query = "How can I contact customer support for help?"

query_embedding = embedding_model.encode(query).tolist()

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=5,
    include=["documents", "distances"]
)

print("Search query:", query)
print("\nTop relevant messages:\n")

for i, (document, distance) in enumerate(
    zip(results["documents"][0], results["distances"][0]),
    start=1
):
    print(f"Result {i}")
    print("Message:", document)
    print("Cosine distance:", round(distance, 4))
    print("-" * 80)
```

### Observation

The similarity search returned five customer support messages relevant to the query. The retrieved messages include support telephone numbers, guidance on contacting a support representative, and requests for additional information. The cosine distances ranged from 0.3301 to 0.3966, with Result 1 having the lowest distance of 0.3301.

### Interpretation

The results indicate that the ChromaDB vector store can retrieve customer support messages related to the meaning of a user's query. The cosine distances provide a measure of dissimilarity, so lower distances indicate closer matches within this search. This confirms that the retrieval component is functioning for the indexed batch of 9,980 records. However, the quality and coverage of retrieval must be assessed further before using the vector store as the complete knowledge base.

These similarity scores indicate retrieval relevance based on textual similarity; they do not independently measure the factual correctness or completeness of the chatbot's generated answers.

# Build the RAG Pipeline

The retriever finds customer support messages relevant to the user's question, and Gemini uses the retrieved information to generate a natural-language response grounded in the available context.

# Evaluate the RAG Pipeline Using Local Metrics

This section evaluates the retrieval performance of the Retrieval-Augmented Generation (RAG) pipeline using local Python-based metrics. It measures the relevance of retrieved customer-support documents and text similarity without requiring additional Gemini API calls.

## Prepare Evaluation Data

Ten test questions will be used to retrieve relevant customer-support documents from the existing ChromaDB vector store. The retrieved contexts will form the evaluation dataset for calculating local retrieval relevance, answer-to-context similarity, and keyword overlap metrics. This step does not generate new chatbot answers or make Gemini API calls.

```python

# Prepare evaluation data using local retrieval only
# This cell does not call Gemini or generate new answers.

evaluation_questions = [
    "How can I contact customer support for help?",
    "How can I get an update about my complaint?",
    "What should I do if I have a problem with a service?",
    "How can I contact a company through direct message?",
    "What contact details should I provide for a complaint update?",
    "How can I ask customer support for assistance?",
    "What information might support need to investigate my issue?",
    "How can I follow up on an unresolved customer support issue?",
    "What should I do if I need help with my account?",
    "How can I communicate my service problem to customer support?"
]

evaluation_data = []

# Retrieve relevant documents locally for each question
for question in evaluation_questions:
    retrieved_docs = retriever.invoke(question)
    contexts = [doc.page_content for doc in retrieved_docs]

    evaluation_data.append({
        "question": question,
        "contexts": contexts
    })

print("Local evaluation data prepared successfully!")
print("Number of test questions:", len(evaluation_data))
print("Retrieved documents for the first question:",
      len(evaluation_data[0]["contexts"]))
```

## Local Retrieval Evaluation Using TF-IDF and Cosine Similarity

This section evaluates the relevance of retrieved customer-support documents using TF-IDF vectorization and cosine similarity. For each test question, the similarity between the question and its five retrieved context documents is calculated. The mean and maximum similarity scores are used as local indicators of retrieval relevance. This evaluation runs locally without making additional Gemini API calls.

```python

# Calculate local retrieval relevance and context similarity

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

evaluation_results = []

for item in evaluation_data:
    question = item["question"]
    contexts = item["contexts"]

    # Calculate similarity between the question and retrieved contexts
    texts = [question] + contexts
    tfidf = TfidfVectorizer(stop_words="english")
    matrix = tfidf.fit_transform(texts)

    similarities = cosine_similarity(
        matrix[0:1], matrix[1:]
    ).flatten()

    evaluation_results.append({
        "question": question,
        "retrieved_documents": len(contexts),
        "mean_retrieval_similarity": float(np.mean(similarities))
        if len(similarities) else 0.0,
        "max_retrieval_similarity": float(np.max(similarities))
        if len(similarities) else 0.0
    })

# Display results for all 10 questions
import pandas as pd

evaluation_results_df = pd.DataFrame(evaluation_results)

print("Local evaluation completed successfully!")
display(evaluation_results_df.round(3))
```

### Observation

The local evaluation was completed successfully for all 10 test questions. For each question, five customer-support context documents were retrieved, resulting in 50 retrieved documents across the evaluation set. The mean retrieval similarity scores ranged from 0.057 to 0.127, while the maximum similarity scores ranged from 0.100 to 0.360. The highest maximum similarity score was 0.360 for the question about contact details required for a complaint update.

### Interpretation

The results show that the retriever returned five context documents for every test question. The TF-IDF cosine-similarity scores indicate varying degrees of textual similarity between the questions and their retrieved contexts. The question about contact details for a complaint update had the highest maximum similarity, while the question about following up on an unresolved customer-support issue had the lowest mean similarity. These results provide an initial indication of retrieval relevance, but they do not establish factual correctness or answer quality. Further inspection of the retrieved documents would help determine whether they adequately address each question.

## Calculate Local Retrieval Evaluation Scores

This section calculates local retrieval-similarity scores for the ten test questions using TF-IDF vectorization and cosine similarity. It reports the mean and maximum similarity between each question and its retrieved customer-support contexts. The evaluation runs locally without additional Gemini API calls.

```python

# Calculate local retrieval-similarity scores

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd

evaluation_results = []

# Calculate similarity for each evaluation question
for item in evaluation_data:
    question = item["question"]
    contexts = item["contexts"]

    if not contexts:
        mean_similarity = 0.0
        max_similarity = 0.0
    else:
        # Convert the question and contexts into TF-IDF vectors
        texts = [question] + contexts
        vectorizer = TfidfVectorizer(stop_words="english")
        tfidf_matrix = vectorizer.fit_transform(texts)

        # Compare the question with each retrieved context
        similarities = cosine_similarity(
            tfidf_matrix[0:1],
            tfidf_matrix[1:]
        ).flatten()

        mean_similarity = float(np.mean(similarities))
        max_similarity = float(np.max(similarities))

    evaluation_results.append({
        "question": question,
        "retrieved_documents": len(contexts),
        "mean_retrieval_similarity": mean_similarity,
        "max_retrieval_similarity": max_similarity
    })

# Display the local evaluation results
evaluation_results_df = pd.DataFrame(evaluation_results)

print("Local retrieval evaluation completed successfully!")
display(evaluation_results_df.round(3))
```

## RAGAS Evaluation

This section evaluates the RAG chatbot using the Faithfulness and Answer Relevancy metrics on the prepared dataset of 10 question-answer examples. Faithfulness assesses whether generated answers are supported by the retrieved contexts, while Answer Relevancy assesses how relevant the answers are to the original questions. The evaluation results will be used to analyze the chatbot's performance.

```python
# Evaluate all 10 examples using the modern RAGAS metrics.
import pandas as pd

ragas_faithfulness_scores = []
ragas_answer_relevancy_scores = []

for i, (question, answer, contexts) in enumerate(
    zip(ragas_questions, ragas_answers, ragas_contexts), start=1
):
    print(f"Evaluating example {i}/10...")

    # Faithfulness requires the question, answer, and retrieved contexts.
    faithfulness_result = await faithfulness_metric.ascore(
        user_input=question,
        response=answer,
        retrieved_contexts=contexts
    )
    ragas_faithfulness_scores.append(faithfulness_result.value)

    # Answer Relevancy requires only the question and answer.
    relevancy_result = await answer_relevancy_metric.ascore(
        user_input=question,
        response=answer
    )
    ragas_answer_relevancy_scores.append(relevancy_result.value)

# Store the results separately from the existing local evaluation.
ragas_results_df = pd.DataFrame({
    "Question": ragas_questions,
    "Faithfulness": ragas_faithfulness_scores,
    "Answer Relevancy": ragas_answer_relevancy_scores
})

print("\nRAGAS evaluation completed.")
display(ragas_results_df)

print("\nAverage Faithfulness:",
      round(ragas_results_df["Faithfulness"].mean(), 4))
print("Average Answer Relevancy:",
      round(ragas_results_df["Answer Relevancy"].mean(), 4))
```
